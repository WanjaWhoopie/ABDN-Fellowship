### Methods to the questions about EEG power and artifacts:

1.  **What frequency range contains most of the EEG power?**
    *   **How to determine:** Look at the generated PSD plot. The highest peaks in the plot typically indicate the frequency ranges with the most power. For EEG, you usually expect to see dominant power in the lower frequency bands (e.g., Delta: 0.5-4 Hz, Theta: 4-8 Hz, Alpha: 8-12 Hz). If you see a large peak in these ranges, that's where most of the EEG power is concentrated.

2.  **Do you see any evidence of line noise?**
    *   **How to determine:** Line noise, also known as power line interference, typically appears as a sharp, narrow peak at a specific frequency, usually 50 Hz or 60 Hz (depending on your geographical location's power grid frequency) and its harmonics (e.g., 100 Hz, 120 Hz). Look for such a distinct, isolated peak in your PSD plot.

3.  **What may cause unusually strong high-frequency power?**
    *   **Causes:** Strong high-frequency power (above 30-40 Hz) in EEG is often indicative of muscle artifacts (EMG). Muscle contractions generate electrical signals that are much faster and higher in frequency than brain activity. Other sources could be electrical interference from nearby equipment or poor electrode contact. In the raw EEG plot, muscle artifacts often appear as irregular, high-frequency, spiky activity.

## **3. Recommended preprocessing steps (and justification)**

### **a. Remove 60 Hz line noise**
- **Method:** Notch filter at 60 Hz (and optionally 120 Hz harmonic).  
- **Reason:** The PSD shows a **sharp, isolated peak at 60 Hz**, the classic signature of power‑line interference. Removing it will clean the spectrum without affecting neural bands.

---

### **b. Apply a band‑pass filter (e.g., 0.5–40 Hz)**
- **Reason:**  
  - EEG power is concentrated below ~20 Hz.  
  - Frequencies above ~40 Hz are mostly noise unless you are specifically analyzing gamma.  
  - A band‑pass filter improves SNR and reduces residual high‑frequency artifacts.

---

### **c. Inspect raw EEG for ocular or movement artifacts**
- **Reason:**  
  - Even though the PSD doesn’t show strong EMG, raw traces often contain eye blinks, saccades, or slow drifts.  
  - These are best identified visually and removed via ICA or regression.

---

### **d. Run ICA (if multichannel data is available)**
- **Reason:**  
  - ICA can isolate blink components, muscle bursts, and residual line noise harmonics.  
  - It is standard practice for cleaning EEG beyond simple filtering.

---

### **e. Re‑reference the data**
- **Method:** Average reference or linked mastoids.  
- **Reason:**  
  - Helps reduce channel‑specific noise.  
  - Produces more stable topographies and cleaner PSDs.

---

### **f. Check electrode impedance and channel quality**
- **Reason:**  
  - Line noise often increases when impedance is high or contact is poor.  
  - Ensuring good electrode contact reduces future contamination.

````python
from pathlib import Path
from mne_bids import BIDSPath

bids_root = Path('./eeg_bids_data')
bids_root.mkdir(exist_ok=True)
print(f'BIDS root directory created at: {bids_root.absolute()}')
````

````python
# Import the necessary function for writing BIDS-compliant data
from mne_bids import write_raw_bids

# Create a BIDSPath object, specifying subject ID, task, run, datatype, and the root directory
# This object defines the file path and metadata for the BIDS dataset
bids_path = BIDSPath(subject='008', task='motor', run='04', datatype='eeg', root=bids_root)

# Write the raw EEG data (loaded as 'raw_data') to a BIDS-compliant format
# 'overwrite=True' allows overwriting existing files
# 'verbose=False' suppresses detailed output during the conversion process
# 'allow_preload=True' permits preloading data into memory if needed
# 'format='EDF'' specifies the output file format for the EEG data
write_raw_bids(raw_data, bids_path=bids_path, overwrite=True, verbose=False, allow_preload=True, format='EDF')

# bids_path.fpath.parent provides the parent directory of the generated BIDS file
print(f'Raw data successfully converted to BIDS format at: {bids_path.fpath.parent}')
````

````python
# Import pandas for data manipulation
import pandas as pd

# Construct the path to the events.tsv file
# This path is inside the BIDS structure we created earlier. We use bids_path.fpath.parent to get the correct directory.
events_tsv_path = bids_path.fpath.parent / f'sub-{bids_path.subject}_task-{bids_path.task}_run-{bids_path.run}_events.tsv'

# Load the events.tsv file into a pandas DataFrame
events_df = pd.read_csv(events_tsv_path, sep='	')

# Print the contents of the events.tsv file
print('Contents of the generated events.tsv file:')
print(events_df)

# Get the number of events present in the loaded file
num_events_tsv = len(events_df)
print(f'\nNumber of events in events.tsv: {num_events_tsv}')

# Extract events from the original raw_data object (from the EDF file)
# MNE automatically converts annotations in the EDF to events
events_from_raw, event_id = mne.events_from_annotations(raw_data)

# Get the number of events extracted directly from the raw EDF data
num_events_raw = len(events_from_raw)
print(f'Number of events from original EDF data: {num_events_raw}')

# Compare the number of events
if num_events_tsv == num_events_raw:
    print('\nBoth the generated events.tsv and the original EDF file contain the same number of events.')
else:
    print('\nThe number of events in events.tsv and the original EDF file differ.')

# Further comparison: check if the onsets are the same
# Convert onsets from seconds in tsv to samples using raw_data.info['sfreq']
events_tsv_onsets_samples = (events_df['onset'] * raw_data.info['sfreq']).round().astype(int)

# Get onsets from the events_from_raw (first column)
events_raw_onsets_samples = events_from_raw[:, 0]

# Check if the onsets in samples are identical
is_onsets_identical = (events_tsv_onsets_samples.values == events_raw_onsets_samples).all()

if is_onsets_identical:
    print('The event onsets (in samples) in events.tsv match the onsets from the original EDF data.')
else:
    print('The event onsets (in samples) in events.tsv DO NOT match the onsets from the original EDF data.')
````
