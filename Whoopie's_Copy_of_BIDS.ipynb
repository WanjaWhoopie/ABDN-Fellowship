```python
input_dir = Path('/content/drive/MyDrive/S001')
bids_root = Path("my1_eeg_bids")
bids_root.mkdir(parents=True, exist_ok=True)
```

# Read the EDF file

```python
run01_file = input_dir / "S001R01.edf"

raw_run01 = mne.io.read_raw_edf(
    run01_file,
    preload=False,
    verbose=True
)

print("=" * 60)
print("RUN 01")
print("=" * 60)

print(raw_run01)
print("Number of channels:", len(raw_run01.ch_names))
print("Sampling frequency:", raw_run01.info["sfreq"])
print("Duration:", raw_run01.times[-1], "seconds")
```

```python

run02_file = input_dir / "S001R02.edf"

raw_run02 = mne.io.read_raw_edf(
    run02_file,
    preload=False,
    verbose=True
)

print("\n" + "=" * 60)
print("RUN 02")
print("=" * 60)

print(raw_run02)
print("Number of channels:", len(raw_run02.ch_names))
print("Sampling frequency:", raw_run02.info["sfreq"])
print("Duration:", raw_run02.times[-1], "seconds")
```

# Create the BIDS path

```python
bids_path_run01 = BIDSPath(
    subject="001",
    task="mentalarithmetic",
    run="01",
    datatype="eeg",
    root=bids_root
)
```

```python
bids_path_run02 = BIDSPath(
    subject="001",
    task="mentalarithmetic",
    run="02",
    datatype="eeg",
    root=bids_root
)
```

```python
print("Run 01:")
print(bids_path_run01.fpath)

print("\nRun 02:")
print(bids_path_run02.fpath)
```

# Convert recordings - Write the recording to BIDS

```python
write_raw_bids( raw=raw_run01,
               bids_path=bids_path_run01,
                format="EDF", overwrite=True, verbose=True )

print("Run 01 converted successfully.")
```

```python
write_raw_bids( raw=raw_run02,
               bids_path=bids_path_run02,
                format="EDF", overwrite=True, verbose=True )

print("Run 02 converted successfully.")
```

```python
# Display all annotations found inside the EDF file
print(raw_run01.annotations)

# Display only the annotation descriptions
print("\nAnnotation descriptions:")
print(raw_run01.annotations.description)

# Display annotation onset times
print("\nAnnotation onsets:")
print(raw_run01.annotations.onset)

# Display annotation durations
print("\nAnnotation durations:")
print(raw_run01.annotations.duration)

```
